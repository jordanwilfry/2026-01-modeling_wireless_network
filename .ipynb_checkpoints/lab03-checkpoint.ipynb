```python
# libraries
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
```

```python
R = 10
N = 1000
```

<strong> Задание:</strong><p> Сгенерировать выборку точек, равномерно распределенных внутри круга
двумя способами:</p>

<strong>1. First method:</strong>

<p>Генерируем точки равномерно в квадрате [−R, R] × [−R, R], описанном вокруг окружности радиуса R. Оставляем только точки, расстояние которых от центра ≤ R, остальные (попавшие в четыре угла квадрата, вне круга) отбрасываем. Это называется метод отбраковки (rejection method): поскольку точки равномерны по всему квадрату, оставшиеся внутри круга точки автоматически равномерны и по кругу.

Так как площадь круга (πR²) составляет около 78.5% площади квадрата (4R²), примерно 1 из 5 сгенерированных точек отбрасывается - поэтому генерируем больше N точек, чтобы наверняка осталось хотя бы N.</p>

```python
x_points = np.random.uniform(-R, R, N*2)
y_points = np.random.uniform(-R, R, N*2)

dist_from_center = np.sqrt(x_points**2 + y_points**2)
inside = dist_from_center <= R

x1 = x_points[inside][:N]
y1 = y_points[inside][:N]

print(f"Method 1 - points kept, {len(x1)}")
```

 <strong>2. second method:</strong>

<p>Точка в полярных координатах задаётся углом θ и расстоянием r от центра. Угол θ равномерен на [0, 2π). Расстояние r НЕ равномерно на [0, R] - если бы было равномерным, точки скапливались бы у центра, т.к. тонкое кольцо у r=0 имеет намного меньшую площадь, чем тонкое кольцо у r=R, а равномерное r даёт в обоих кольцах одинаковое число точек. Чтобы точки были равномерны по площади, нужно:

r = R·√u, где u равномерно на [0, 1]

Затем переводим обратно в декартовы координаты: x = r·cos(θ), y = r·sin(θ)</p>

```python
theta = np.random.uniform(0, 2 * np.pi, N)
U = np.random.uniform(0, 1, N)
r = R * np.sqrt(U)

x2 = r * np.cos(theta)
y2 = r * np.sin(theta)

print(f"Method 2 - points generated, {len(x2)}")
```

</br><strong>Задание 1:</strong> Создать рисунок, иллюстрирующий расположение точек сгенерированной
выборки внутри окружности;

```python
fig, axes = plt.subplots(1, 2)

axes[0].scatter(x1, y1, s=5)
axes[0].set_title("Methos 1")
axes[0].set_aspect("equal")

axes[1].scatter(x2, y2, s=5, color="orange")
axes[1].set_title("Methos 2")
axes[1].set_aspect("equal")

plt.show()
```

</br><strong>Задание 2:</strong> Найти выборочные средние координат точек и их дисперсию;

<p>По симметрии теоретическое среднее x и y равно 0 (круг центрирован в начале координат). Теоретическая дисперсия каждой координаты равна R²/4 — известный результат для равномерного распределения по кругу радиуса R.</p>

```python
print (f"Method 1: mean x1 = {np.mean(x1)}, mean y1 = {np.mean(y1)}")
print (f"Method 1: var x1 = {np.var(x1)}, var y1 = {np.var(y1)}")

print (f"\nMethod 2: mean x2 = {np.mean(x2)}, mean y2 = {np.mean(y2)}")
print (f"Method 2: var x2 = {np.var(x2)}, var y2 = {np.var(y2)}")

print(f"\nTheory: mean = 0, var  = {R**2/4}")
```

</br><strong>Задание 3:</strong> Построить график плотности распределения расстояния от случайной равномерно
распределенной точки в круге до фиксированной точки лежащей вне окружности
(к примеру, с координатами X=20, Y=0).

<p>
    Для каждой точки круга считаем евклидово расстояние до фиксированной внешней точки (20, 0), затем строим плотность этих расстояний — и в виде гистограммы, и в виде гладкой кривой KDE.
</p>

```python
Px, Py = 20, 0

dist_to_P_1 = np.sqrt((x1 - Px)**2 + (y1 - Py)**2)
dist_to_P_2 = np.sqrt((x2 - Px)**2 + (y2 - Py)**2)


plt.hist(dist_to_P_1, bins=30, density=True, alpha=0.4, label="Method 1 histogram")
plt.hist(dist_to_P_2, bins=30, density=True, alpha=0.4, label="Method 2 histogram")

grid = np.linspace(min(dist_to_P_1.min(), dist_to_P_2.min()), max(dist_to_P_1.max(), dist_to_P_2.max()), 200)

kde_1 = gaussian_kde(dist_to_P_1)
kde_2 = gaussian_kde(dist_to_P_2)

plt.plot(grid, kde_1(grid), label="method 1 KDE", linewidth = 2)
plt.plot(grid, kde_2(grid), label="method 2 KDE", linewidth = 2)

plt.title("Distance to point (20, 0)")
plt.xlabel("Distance)")
plt.ylabel("density)")
plt.legend()
plt.show()
```

</br><strong>Задание 4:</strong> Построить график плотности распределения расстояния между двумя
случайными точками, равномерно расположенными внутри круга.

```python
# Second independent sample inside the circle, using Method 2
theta_random = np.random.uniform(0, 2 * np.pi, N)
r_random = R * np.sqrt(np.random.uniform(0, 1, N))

x_random = r_random * np.cos(theta_random)
y_random = r_random * np.sin(theta_random)

dist_btw_points = np.sqrt((x2 - x_random)**2 + (y2 - y_random)**2)

plt.hist(dist_btw_points, bins=30, density=True, alpha=0.4, label="Histogram")

grid_random = np.linspace(dist_btw_points.min(), dist_btw_points.max(), 200)

kde_random = gaussian_kde(dist_btw_points)

plt.plot(grid_random, kde_random(grid_random), label="KDE", linewidth = 2)

plt.title("Distance between 2 random points in the circle")
plt.xlabel("Distance)")
plt.ylabel("density)")
plt.legend()
plt.show()

print(f"\nMean distance between two pointts is: {np.mean(dist_btw_points)}")
```
