<strong>Задание 1</strong> Построить матрицу c количеством строк 10 и столбцов 2, переменные 
заполняются случайными числами с равномерным распределением. 
Полученные сгенерированные случайные числа представить на графике в 
виде точек.

```python
#libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
matrix = np.random.uniform(0, 1, (10, 2))
print (matrix)
```

```python
plt.scatter(matrix[:, 0], matrix[:,1])
plt.title("Задание 1")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

<strong>Задание 2</strong> Сгенерировать 1000 случайных чисел с любым известным распределением 
и построить их гистограмму, математическое ожидание, дисперсию.

```python
data = np.random.uniform(0, 1, 1000)

print("Mean/ математическое ожидание:", np.mean(data))
print("Variance/ дисперсию:", np.var(data))
```

```python
plt.hist(data, bins=50)
plt.title("Задание 2")
plt.xlabel("value")
plt.ylabel("count")
plt.show()
```

<strong>Задание 3</strong> Сгенерировать случайную точку, равномерно распределенную в квадрате 
со стороной а

```python
a = 10
x = np.random.uniform(0, a)
y = np.random.uniform(0, a)

print(f"точку x:{x}, y:{y}")
```

```python
plt.plot([0,a,a,0,0],[0, 0, a, a, 0], color="black")
plt.scatter(x, y)
plt.title("Задание 3")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```
